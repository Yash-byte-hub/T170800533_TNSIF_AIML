```python
import pandas as pd

# Load the CSV file
df = pd.read_csv("data.csv")

# Display the first 5 rows
print("--- First 5 Rows ---")
display(df.head())

# Display the last 5 rows
print("\n--- Last 5 Rows ---")
display(df.tail())
```

```python
# 1. Display total number of rows and columns
print("--- Total Rows & Columns ---")
print(df.shape)

# 2. Display data types of each column
print("\n--- Data Types ---")
print(df.dtypes)
```

```python
# 1. Count missing values per column
missing_counts = df.isnull().sum()

# 2. Filter and display only columns that have at least one missing value
print("--- Columns with Missing Values & Their Counts ---")
print(missing_counts[missing_counts > 0])
```

```python
# Loop through each column in the DataFrame
for col in df.columns:
    # Check if the column is numerical (integers or decimals)
    if df[col].dtype in ['int64', 'float64']:
        mean_value = df[col].mean()
        df[col] = df[col].fillna(mean_value)

    # Check if the column is categorical (text or objects)
    elif df[col].dtype == 'object':
        # .mode()[0] gets the most frequent value in that column
        mode_value = df[col].mode()[0]
        df[col] = df[col].fillna(mode_value)

# Verify that all missing values are filled
print("--- Remaining Missing Values ---")
print(df.isnull().sum())
```

```python
# 1. Check how many duplicates exist before removing them
print(f"Number of duplicate rows before: {df.duplicated().sum()}")

# 2. Remove duplicate rows
df = df.drop_duplicates()

# 3. Verify they are gone
print(f"Number of duplicate rows after: {df.duplicated().sum()}")
print(f"New total dataset size: {df.shape}")
```

```python
# Print all column names exactly as they are written
print(df.columns.tolist())
```

```python
# 1. Rename all columns to lowercase
df.columns = df.columns.str.lower()

# 2. Display the new column names
print("--- Updated Column Names ---")
print(df.columns.tolist())
```

```python
# 1. Automatically find and convert applicable float columns to integers
for col in df.columns:
    if df[col].dtype == 'float64':
        # Check if the column has no missing values
        if df[col].isnull().sum() == 0:
            df[col] = df[col].astype('int64')
            print(f"Converted column to integer: '{col}'")

# 2. Verify your updated data types
print("\n--- Current Data Types ---")
print(df.dtypes)
```

```python
# 1. Choose the specific columns you want to fill
columns_to_fill = ['math_marks', 'science_marks', 'english_marks']

# 2. Fill missing values in these columns with a specific value (e.g., 0)
df[columns_to_fill] = df[columns_to_fill].fillna(0)

# 3. Verify that these columns no longer have missing values
print("--- Remaining Missing Values ---")
print(df[columns_to_fill].isnull().sum())
```

```python
# 1. Define the columns you want to transform
marks_columns = ['math_marks', 'science_marks', 'english_marks']

# 2. Increase the marks by 10%
df[marks_columns] = df[marks_columns] * 1.10

# 3. View a few rows to confirm the updated scores
print("--- Updated Marks (First 5 Rows) ---")
display(df[marks_columns].head())
```

```python
# Multiply by 10%, round to the nearest whole number, and convert to integer
df[marks_columns] = (df[marks_columns] * 1.10).round(0).astype(int)
```

```python
# Convert all text in the column to UPPERCASE
df['department'] = df['department'].str.upper()

# Verify the changes
print(df['department'].unique())
```

```python
import numpy as np

# Define your passing threshold (e.g., 40 marks)
passing_score = 40

# 1. Option A: Pass/Fail based on the average score across all subjects
average_marks = (df['math_marks'] + df['science_marks'] + df['english_marks']) / 3

df['status'] = np.where(average_marks >= passing_score, 'Pass', 'Fail')

# 2. Option B: Pass ONLY if they score above the threshold in EVERY subject
# df['status'] = np.where(
#     (df['math_marks'] >= passing_score) &
#     (df['science_marks'] >= passing_score) &
#     (df['english_marks'] >= passing_score),
#     'Pass', 'Fail'
# )

# Check your results
print("--- Status Counts ---")
print(df['status'].value_counts())

print("\n--- Quick View ---")
display(df[['math_marks', 'science_marks', 'english_marks', 'status']].head())
```

```python
# 1. Filter students with marks > 80 (e.g., in Math)
high_math_scorers = df[df['math_marks'] > 80]

print("--- Students with Math Marks > 80 ---")
display(high_math_scorers[['name', 'department', 'math_marks']].head())


# 2. Filter data where Rainfall > 10
# (Uncomment and use this if you have a separate weather dataset loaded as 'df_weather')
# heavy_rain = df_weather[df_weather['rainfall'] > 10]


# 3. Alternative: Filter students who scored > 80 in ALL three subjects
excellent_students = df[(df['math_marks'] > 80) &
                        (df['science_marks'] > 80) &
                        (df['english_marks'] > 80)]

print("\n--- Students with > 80 in All Subjects ---")
display(excellent_students[['name', 'math_marks', 'science_marks', 'english_marks']].head())
```
