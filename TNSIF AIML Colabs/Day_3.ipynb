```python
num = 12345
reversed_num = 0

# Track the sign for negative numbers
sign = -1 if num < 0 else 1
temp = abs(num)

while temp > 0:
    last_digit = temp % 10          # Get the last digit
    reversed_num = (reversed_num * 10) + last_digit  # Shift left and add digit
    temp //= 10                     # Remove the last digit

# Restore the negative sign if necessary
reversed_num *= sign

print(f"Reversed number: {reversed_num}")
```

```python
num = 29
is_prime = True

# Numbers less than or equal to 1 are not prime
if num <= 1:
    is_prime = False
else:
    # Check factors from 2 up to the square root of num
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break  # Stop the loop immediately if a factor is found

# Print the result
if is_prime:
    print(f"{num} is a prime number.")
else:
    print(f"{num} is not a prime number.")
```

```python
def is_palindrome_math(num: int) -> bool:
    # Negative numbers cannot be palindromes due to the minus sign
    if num < 0:
        return False

    original_num = num
    reversed_num = 0

    while num > 0:
        last_digit = num % 10
        reversed_num = (reversed_num * 10) + last_digit
        num //= 10  # Remove the last digit

    # Return True if they match, otherwise False
    return original_num == reversed_num

# Test cases
print(is_palindrome_math(121))    # Returns: True
print(is_palindrome_math(-121))   # Returns: False
print(is_palindrome_math(1234))   # Returns: False
```

```python
def sum_of_digits_math(num: int) -> int:
    # Handle negative numbers by converting to positive
    num = abs(num)
    digit_sum = 0

    while num > 0:
        digit_sum += num % 10  # Extract the last digit and add to sum
        num //= 10             # Remove the last digit

    return digit_sum

# Test cases
print(sum_of_digits_math(1234))   # Returns: 10 (1 + 2 + 3 + 4)
print(sum_of_digits_math(-56))    # Returns: 11 (5 + 6)
print(sum_of_digits_math(0))      # Returns: 0
```

```python
# Create a list of items
items = [10, 20, 30, 40, 50]

# Initialize pointers at the first and last indices
left = 0
right = len(items) - 1

# Swap elements until the pointers meet in the middle
while left < right:
    # Pythonic shortcut to swap two variables
    items[left], items[right] = items[right], items[left]

    # Move the pointers inward
    left += 1
    right -= 1

print(f"Reversed list (In-place): {items}")
```

```python
# Create two initial sets
set_a = {1, 2, 3, 4, 5}
set_b = {4, 5, 6, 7, 8}

# Perform Union (all unique elements from both sets)
union_result = set_a.union(set_b)

# Perform Intersection (only elements present in both sets)
intersection_result = set_a.intersection(set_b)

print(f"Set A: {set_a}")
print(f"Set B: {set_b}")
print(f"Union: {union_result}")          # Output: {1, 2, 3, 4, 5, 6, 7, 8}
print(f"Intersection: {intersection_result}")  # Output: {4, 5}
```

```python
scores = {"Alice": 85, "Bob": 92, "Charlie": 78, "Diana": 95}

# Sort in Ascending Order (Lowest to Highest value)
sorted_ascending = dict(sorted(scores.items(), key=lambda item: item[1]))

# Sort in Descending Order (Highest to Lowest value)
sorted_descending = dict(sorted(scores.items(), key=lambda item: item[1], reverse=True))

print("Ascending:", sorted_ascending)
# Output: {'Charlie': 78, 'Alice': 85, 'Bob': 92, 'Diana': 95}

print("Descending:", sorted_descending)
# Output: {'Diana': 95, 'Bob': 92, 'Alice': 85, 'Charlie': 78}
```

```python
class Student:
    # The constructor method to initialize attributes
    def __init__(self, name: str, age: int):
        self.name = name  # Instance variable for student's name
        self.age = age    # Instance variable for student's age

    # A custom method to display student details
    def print_details(self):
        print(f"Student Details -> Name: {self.name}, Age: {self.age}")


# --- Using the Class ---

# Step 1: Create (instantiate) two student objects
student1 = Student("Alice", 20)
student2 = Student("Bob", 22)

# Step 2: Call the method to print their details
student1.print_details()
student2.print_details()
```

```python
class Rectangle:
    # Constructor to initialize dimensions
    def __init__(self, length: float, width: float):
        self.length = length
        self.width = width

    # Method to calculate area
    def calculate_area(self) -> float:
        return self.length * self.width

    # Method to calculate perimeter
    def calculate_perimeter(self) -> float:
        return 2 * (self.length + self.width)


# --- Using the Class ---

# Create a rectangle instance with length 10 and width 5
rect = Rectangle(10.0, 5.0)

# Compute values using class methods
area = rect.calculate_area()
perimeter = rect.calculate_perimeter()

# Print the results
print(f"Rectangle Dimensions: {rect.length} x {rect.width}")
print(f"Area: {area}")           # Output: 50.0
print(f"Perimeter: {perimeter}")   # Output: 30.0
```

```python
class Employee:
    # Constructor to initialize employee details
    def __init__(self, name: str, salary: float):
        self.name = name
        self.salary = salary

    # Method to increase salary by a given percentage
    def give_raise(self, percentage: float):
        if percentage > 0:
            increase_amount = self.salary * (percentage / 100)
            self.salary += increase_amount
            print(f"Success: {self.name} received a {percentage}% raise (+₹{increase_amount:,.2f})")
        else:
            print("Error: Raise percentage must be greater than 0.")

    # Method to display current employee status
    def display_info(self):
        print(f"Employee: {self.name} | Current Salary: ₹{self.salary:,.2f}")


# --- Using the Class ---

# Step 1: Create an employee instance
emp = Employee("Rajesh Kumar", 50000.0)
emp.display_info()  # Initial salary

# Step 2: Give the employee a 10% raise
emp.give_raise(10.0)

# Step 3: Check the updated details
emp.display_info()
```

```python
class ShoppingCart:
    # Constructor to initialize an empty cart dictionary
    def __init__(self):
        self.cart = {}  # Format will be -> {"item_name": price}

    # Method to add an item and its price to the cart
    def add_item(self, item_name: str, price: float):
        if price >= 0:
            self.cart[item_name] = price
            print(f"Added to cart: {item_name} (₹{price:,.2f})")
        else:
            print("Error: Price cannot be negative.")

    # Method to calculate the total bill amount
    def calculate_total(self) -> float:
        # Sum up all the values (prices) in the dictionary
        total_bill = sum(self.cart.values())
        return total_bill

    # Method to display items currently in the cart
    def display_cart(self):
        if not self.cart:
            print("Your shopping cart is empty.")
        else:
            print("\n--- Current Items in Cart ---")
            for item, price in self.cart.items():
                print(f"- {item}: ₹{price:,.2f}")
            print(f"Total Bill: ₹{self.calculate_total():,.2f}\n")


# --- Using the Class ---

# Step 1: Create a shopping cart instance
my_cart = ShoppingCart()

# Step 2: Add some products
my_cart.add_item("Wireless Mouse", 1250.00)
my_cart.add_item("Mechanical Keyboard", 3499.50)
my_cart.add_item("USB-C Cable", 450.00)

# Step 3: Display the cart list and total bill
my_cart.display_cart()
```

```python
class Library:
    # Constructor to initialize an empty list for books
    def __init__(self):
        self.books = []  # List to store book titles as strings

    # Method to add a book to the library
    def add_book(self, book_title: str):
        self.books.append(book_title)
        print(f"Added: '{book_title}' has been added to the library.")

    # Method to remove a book from the library
    def remove_book(self, book_title: str):
        if book_title in self.books:
            self.books.remove(book_title)
            print(f"Removed: '{book_title}' has been removed from the library.")
        else:
            print(f"Error: '{book_title}' is not available in the library.")

    # Method to display all books currently in the library
    def display_books(self):
        if not self.books:
            print("\nThe library is currently empty.")
        else:
            print(f"\n--- Library Catalog ({len(self.books)} books) ---")
            for index, book in enumerate(self.books, start=1):
                print(f"{index}. {book}")
            print("---------------------------------------\n")


# --- Using the Class ---

# Step 1: Create a library instance
city_library = Library()

# Step 2: Add books to the collection
city_library.add_book("The Hobbit")
city_library.add_book("To Kill a Mockingbird")
city_library.add_book("1984")

# Step 3: Display current collection
city_library.display_books()

# Step 4: Remove a book and test the error handling
city_library.remove_book("1984")          # Successful removal
city_library.remove_book("Harry Potter")  # Triggers error message

# Step 5: Display the finalized catalog
city_library.display_books()
```
